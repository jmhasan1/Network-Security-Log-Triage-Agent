````python
from pathlib import Path
import sys
import numpy as np
import pandas as pd

def find_project_root(start=None):
    start = Path.cwd() if start is None else Path(start).resolve()
    for candidate in [start, *start.parents]:
        if (candidate / "src").exists() and (candidate / "data").exists():
            return candidate
    raise FileNotFoundError("Open this notebook from inside the project repository.")

PROJECT_ROOT = find_project_root()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
print("Project root:", PROJECT_ROOT)
````

````python
RAW_DATA_PATH = PROJECT_ROOT / "data" / "raw" / "phisingData.csv"
TARGET_COLUMN = "Result"
df = pd.read_csv(RAW_DATA_PATH)
print(df.shape)
display(df.head())
````

````python
overview = pd.DataFrame({
    "dtype": df.dtypes.astype(str),
    "missing": df.isna().sum(),
    "unique": df.nunique(dropna=False),
})
display(overview)
````

````python
if TARGET_COLUMN in df.columns:
    display(df[TARGET_COLUMN].value_counts(dropna=False).to_frame("count"))
    print(df[TARGET_COLUMN].value_counts(normalize=True).rename("ratio"))
````

````python
numeric_df = df.select_dtypes(include=np.number)
display(numeric_df.describe().T)
display(numeric_df.corr().round(3))
````
